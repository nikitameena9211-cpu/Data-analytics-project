```python
#1Customer feature
customer_feature =analysis_df.groupby('Customer_ID').agg(
    Total_orders = ('Order_ID','nunique'),
    Total_spent = ('Total_Amount','sum'),
    Total_quantity =('Quantity','sum')
    )
customer_feature
```

```python
#RFM Feature 
reference_date = analysis_df['Order_Date'].max() + pd.Timedelta(days=1)

rfm = analysis_df.groupby('Customer_ID').agg(
    recency=('Order_Date', lambda x: (reference_date - x.max()).days),
    frequency=('Order_ID', 'count'),
    monetary=('Total_Amount', 'sum')
).reset_index()
reference_date                                           
```

```python
# RFM Score
R_score = pd.qcut(
    rfm['recency'],
    5,
    labels=[5, 4, 3, 2, 1])
F_score = pd.qcut(
    rfm['frequency'],
    5,
    labels=[1, 2, 3, 4, 5])
M_score = pd.qcut(
    rfm['monetary'],
    5,
    labels=[1, 2, 3, 4, 5])
rfm['R_score'] = R_score.astype(int)
rfm['F_score'] = F_score.astype(int)
rfm['M_score'] = M_score.astype(int)
rfm['RFM_Score'] = (
    rfm['R_score'].astype(str) +
    rfm['F_score'].astype(str) +
    rfm['M_score'].astype(str)
)
```

```python
rfm['RFM_Score']
```

```python
# Individual scores
rfm['R_score'] = R_score.astype(int)
rfm['F_score'] = F_score.astype(int)
rfm['M_score'] = M_score.astype(int)
# 3-digit RFM score
rfm['RFM_Score'] = (
    rfm['R_score'].astype(str) +
    rfm['F_score'].astype(str) +
    rfm['M_score'].astype(str)
)
# Total RFM score
rfm['RFM_Total'] = (
    rfm['R_score'] +
    rfm['F_score'] +
    rfm['M_score']
)
# Customer segmentation
def customer_segment(score):
    if score >= 13:
        return 'Champions'
    elif score >= 10:
        return 'Loyal Customers'
    elif score >= 7:
        return 'Potential Loyal'
    elif score >= 5:
        return 'At Risk'
    else:
        return 'Lost Customers'

rfm['Customer_Segment'] = rfm['RFM_Total'].apply(customer_segment)
rfm['Customer_Segment']
```

#Key insights and Recommandations
Insight 1 — Category Performance
Electronics is the dominant revenue-generating category, contributing approximately 72.61% of total revenue. 
It also records the highest customer count, transaction volume, quantity sold, and average order value, making it the primary revenue driver.
Recommendation:
Maintain strong inventory availability and focus on retaining high-value Electronics customers.
    
Insight 2 — Monthly Performance
June recorded the highest overall monthly revenue at approximately ₹708.7M, supported by higher customer count, transactions, and average order value. 
However, the highest-performing month varies across years, so June should not be considered a consistent seasonal peak.
Recommendation:
Analyze the drivers of each year's peak month before designing seasonal campaigns.

Insight 3 — Geographic Performance
UP, Rajasthan, and Haryana are among the highest-revenue states and also have larger customer bases.
Customer volume appears to be an important contributor to their strong revenue performance.
Recommendation:
Maintain customer acquisition and retention efforts in high-performing states while investigating lower-customer states for growth opportunities.

Insight 4-Customer Segment Performance
Loyal Customers generate the highest revenue among the RFM segments, followed by Champions and Potential Loyal customers. 
At Risk and Lost Customers contribute comparatively lower revenue.
Recommendation:
Strengthen retention strategies for Loyal and Champion customers while using targeted engagement campaigns to move Potential Loyal customers
toward higher-value segments.
